```python
import json
import statistics
```

```python
file = "dataset_final.jsonl"
```

```python
word_lengths = []
tokens = []
tokens_english = []
```

```python
with open(file, 'r') as f:
    for line in f:
        data = json.loads(line)
        text = data.get("description", "")

        words = len(text.split())
        word_lengths.append(words)

        est_tokens = words * 1.3
        tokens.append(est_tokens)

        tokens_engl = est_tokens * 1.1
        tokens_english.append(tokens_engl)
```

```python
print("records:", len(word_lengths))
print("words avg:", round(statistics.mean(word_lengths), 2))
print("words median:", statistics.median(word_lengths))
print("Min words:", min(word_lengths))
print("Max words:", max(word_lengths))

print("\nEstimate after translation:")
print("Avg token:", round(statistics.mean(tokens_english)))
print("tot estimated token:", round(sum(tokens_english)))
```

```python
from collections import Counter
```

```python
countries = []

with open(file, "r", encoding="utf-8") as f:
    for line in f:
        record = json.loads(line)
        country = record.get("country", "Unknown")
        countries.append(country)

counter = Counter(countries)
```

```python
print("Distribuzione nazioni:")
for country, count in counter.most_common():
    print(f"{country}: {count}")
```
